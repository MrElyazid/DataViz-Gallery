# Enhanced World Population Analysis

This notebook demonstrates enhanced visualization techniques for the World Population dataset using custom utility functions.

Dataset link: https://www.kaggle.com/datasets/iamsouravbanerjee/world-population-dataset

```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Import our custom utility functions
from world_pop_utils import (
    load_world_population_data,
    plot_top_countries_bar,
    plot_population_pie,
    plot_population_growth,
    plot_population_density_scatter,
    plot_continent_population_comparison
)
```

## 1. Load and Explore the Dataset

```python
# Load the World Population dataset
df = load_world_population_data('world_population.csv')

# Display the first few rows
print("First 5 rows of the World Population dataset:")
display(df.head())

# Display basic information about the dataset
print("\nDataset information:")
display(df.info())

# Display basic statistics
print("\nBasic statistics:")
display(df.describe())

# Check for missing values
print("\nMissing values:")
display(df.isnull().sum())
```

## 2. Top 10 Countries by Population (2022)

```python
# Create an enhanced bar chart of the top 10 countries by population
top_countries_fig = plot_top_countries_bar(df, n=10, year='2022 Population')
plt.show()
```

## 3. Population Distribution Pie Chart

```python
# Create an enhanced pie chart showing population distribution
pie_chart_fig = plot_population_pie(df, n=10, year='2022 Population')
plt.show()
```

## 4. Population Growth Over Time (1970-2022)

```python
# Create a line chart showing population growth for top 5 countries
growth_fig = plot_population_growth(df, n=5)
plt.show()
```

## 5. Population vs Population Density

```python
# Create a scatter plot of population vs density
density_fig = plot_population_density_scatter(df)
plt.show()
```

## 6. Population by Continent Comparison (1970 vs 2022)

```python
# Create a grouped bar chart comparing population by continent
continent_fig = plot_continent_population_comparison(df, years=['1970 Population', '2022 Population'])
plt.show()
```

## 7. Growth Rate Analysis

```python
# Analyze growth rates
# Get top 20 countries by growth rate
top_growth = df.nlargest(20, 'Growth Rate')[['Country/Territory', 'Continent', 'Growth Rate', '2022 Population']]

# Create a horizontal bar chart
plt.figure(figsize=(12, 10))

# Create a colormap based on continent
continents = top_growth['Continent'].unique()
continent_colors = dict(zip(continents, sns.color_palette("viridis", len(continents))))
bar_colors = [continent_colors[cont] for cont in top_growth['Continent']]

# Create the horizontal bar chart
bars = plt.barh(top_growth['Country/Territory'], 
               (top_growth['Growth Rate'] - 1) * 100,  # Convert to percentage
               color=bar_colors,
               edgecolor='black',
               linewidth=1,
               alpha=0.8)

# Add data labels
for i, bar in enumerate(bars):
    width = bar.get_width()
    population = top_growth['2022 Population'].iloc[i]
    pop_str = f"{population/1e6:.1f}M" if population < 1e9 else f"{population/1e9:.2f}B"
    plt.text(width + 0.1, bar.get_y() + bar.get_height()/2, 
             f"{width:.1f}% ({pop_str})",
             va='center', fontsize=9, fontweight='bold')

# Add title and labels
plt.title('Top 20 Countries by Population Growth Rate', fontsize=18, fontweight='bold', pad=20)
plt.xlabel('Annual Growth Rate (%)', fontsize=14, labelpad=10)
plt.ylabel('Country', fontsize=14, labelpad=10)

# Add grid lines
plt.grid(axis='x', linestyle='--', alpha=0.7)

# Add a subtle background color
plt.gca().set_facecolor('#f8f9fa')

# Create a legend for continents
legend_elements = [plt.Line2D([0], [0], marker='s', color='w', 
                             markerfacecolor=color, markersize=10, 
                             label=continent)
                  for continent, color in continent_colors.items()]

plt.legend(handles=legend_elements, title='Continent', title_fontsize=12, fontsize=10,
          loc='lower right', frameon=True, facecolor='white', edgecolor='gray')

plt.tight_layout()
plt.show()

# Display the data
print("Top 20 Countries by Growth Rate:")
display(top_growth)
```

## 8. Population Density Analysis

```python
# Get top 15 countries by population density
top_density = df.nlargest(15, 'Density (per km²)')[['Country/Territory', 'Continent', 'Density (per km²)', '2022 Population', 'Area (km²)']]

# Create a horizontal bar chart
plt.figure(figsize=(12, 8))

# Create a colormap based on continent
continents = top_density['Continent'].unique()
continent_colors = dict(zip(continents, sns.color_palette("viridis", len(continents))))
bar_colors = [continent_colors[cont] for cont in top_density['Continent']]

# Create the horizontal bar chart
bars = plt.barh(top_density['Country/Territory'], 
               top_density['Density (per km²)'],
               color=bar_colors,
               edgecolor='black',
               linewidth=1,
               alpha=0.8)

# Add data labels
for i, bar in enumerate(bars):
    width = bar.get_width()
    area = top_density['Area (km²)'].iloc[i]
    plt.text(width + 100, bar.get_y() + bar.get_height()/2, 
             f"Area: {area:,} km²",
             va='center', fontsize=9, fontweight='bold')

# Add title and labels
plt.title('Top 15 Countries by Population Density', fontsize=18, fontweight='bold', pad=20)
plt.xlabel('Population Density (people per km²)', fontsize=14, labelpad=10)
plt.ylabel('Country', fontsize=14, labelpad=10)

# Add grid lines
plt.grid(axis='x', linestyle='--', alpha=0.7)

# Add a subtle background color
plt.gca().set_facecolor('#f8f9fa')

# Create a legend for continents
legend_elements = [plt.Line2D([0], [0], marker='s', color='w', 
                             markerfacecolor=color, markersize=10, 
                             label=continent)
                  for continent, color in continent_colors.items()]

plt.legend(handles=legend_elements, title='Continent', title_fontsize=12, fontsize=10,
          loc='lower right', frameon=True, facecolor='white', edgecolor='gray')

plt.tight_layout()
plt.show()

# Display the data
print("Top 15 Countries by Population Density:")
display(top_density)
```

## 9. Correlation Analysis

```python
# Select numerical columns for correlation analysis
numerical_cols = [
    '2022 Population', '2020 Population', '2015 Population', '2010 Population',
    '2000 Population', '1990 Population', '1980 Population', '1970 Population',
    'Area (km²)', 'Density (per km²)', 'Growth Rate', 'World Population Percentage'
]

# Calculate correlation matrix
corr_matrix = df[numerical_cols].corr()

# Create a heatmap of the correlation matrix
plt.figure(figsize=(14, 12))
sns.heatmap(corr_matrix, annot=True, cmap='viridis', vmin=-1, vmax=1, 
            linewidths=0.5, cbar_kws={'label': 'Correlation Coefficient'},
            fmt='.2f', annot_kws={"size": 8})
plt.title('Correlation Matrix of World Population Features', fontsize=18, fontweight='bold', pad=20)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

## 10. Continent Analysis

```python
# Group by continent and calculate statistics
continent_stats = df.groupby('Continent').agg({
    'Country/Territory': 'count',
    '2022 Population': 'sum',
    'Area (km²)': 'sum',
    'Density (per km²)': 'mean',
    'Growth Rate': 'mean'
}).reset_index()

# Rename columns for clarity
continent_stats = continent_stats.rename(columns={
    'Country/Territory': 'Number of Countries',
    '2022 Population': 'Total Population',
    'Area (km²)': 'Total Area (km²)',
    'Density (per km²)': 'Average Density (per km²)',
    'Growth Rate': 'Average Growth Rate'
})

# Calculate population percentage
continent_stats['Population Percentage'] = (continent_stats['Total Population'] / continent_stats['Total Population'].sum()) * 100

# Calculate area percentage
continent_stats['Area Percentage'] = (continent_stats['Total Area (km²)'] / continent_stats['Total Area (km²)'].sum()) * 100

# Sort by total population
continent_stats = continent_stats.sort_values('Total Population', ascending=False)

# Display the continent statistics
print("Continent Statistics:")
display(continent_stats)

# Create a figure with 2 subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 8))

# Create a pie chart for population percentage
colors = sns.color_palette("viridis", len(continent_stats))
wedges, texts, autotexts = ax1.pie(
    continent_stats['Population Percentage'],
    labels=None,
    autopct=None,
    colors=colors,
    wedgeprops=dict(width=0.5, edgecolor='white', linewidth=2),
    startangle=90,
    shadow=True
)

# Add a white circle at the center
centre_circle = plt.Circle((0, 0), 0.3, fc='white', edgecolor='gray', linewidth=1)
ax1.add_patch(centre_circle)

# Add custom labels with lines
bbox_props = dict(boxstyle="round,pad=0.3", fc="white", ec="gray", alpha=0.8)
kw = dict(arrowprops=dict(arrowstyle="-", color="gray", linewidth=1),
         bbox=bbox_props, zorder=0, va="center")

for i, p in enumerate(wedges):
    ang = (p.theta2 - p.theta1)/2. + p.theta1
    y = np.sin(np.deg2rad(ang))
    x = np.cos(np.deg2rad(ang))
    
    # Calculate label position
    horizontalalignment = {-1: "right", 1: "left"}[int(np.sign(x))]
    connectionstyle = f"angle,angleA=0,angleB={ang}"
    kw["arrowprops"].update({"connectionstyle": connectionstyle})
    
    # Create the label text
    continent = continent_stats['Continent'].iloc[i]
    percentage = continent_stats['Population Percentage'].iloc[i]
    population = continent_stats['Total Population'].iloc[i]
    pop_str = f"{population/1e9:.2f}B" if population >= 1e9 else f"{population/1e6:.1f}M"
    
    label_text = f"{continent}\n{pop_str} ({percentage:.1f}%)"
    
    # Position the label
    ax1.annotate(label_text, xy=(x*1.1, y*1.1), xytext=(1.5*x, 1.5*y),
               horizontalalignment=horizontalalignment, **kw)

ax1.set_title('Population Distribution by Continent', fontsize=16, fontweight='bold')

# Create a pie chart for area percentage
wedges, texts, autotexts = ax2.pie(
    continent_stats['Area Percentage'],
    labels=None,
    autopct=None,
    colors=colors,
    wedgeprops=dict(width=0.5, edgecolor='white', linewidth=2),
    startangle=90,
    shadow=True
)

# Add a white circle at the center
centre_circle = plt.Circle((0, 0), 0.3, fc='white', edgecolor='gray', linewidth=1)
ax2.add_patch(centre_circle)

# Add custom labels with lines
for i, p in enumerate(wedges):
    ang = (p.theta2 - p.theta1)/2. + p.theta1
    y = np.sin(np.deg2rad(ang))
    x = np.cos(np.deg2rad(ang))
    
    # Calculate label position
    horizontalalignment = {-1: "right", 1: "left"}[int(np.sign(x))]
    connectionstyle = f"angle,angleA=0,angleB={ang}"
    kw["arrowprops"].update({"connectionstyle": connectionstyle})
    
    # Create the label text
    continent = continent_stats['Continent'].iloc[i]
    percentage = continent_stats['Area Percentage'].iloc[i]
    area = continent_stats['Total Area (km²)'].iloc[i]
    area_str = f"{area/1e6:.2f}M km²"
    
    label_text = f"{continent}\n{area_str} ({percentage:.1f}%)"
    
    # Position the label
    ax2.annotate(label_text, xy=(x*1.1, y*1.1), xytext=(1.5*x, 1.5*y),
               horizontalalignment=horizontalalignment, **kw)

ax2.set_title('Land Area Distribution by Continent', fontsize=16, fontweight='bold')

# Add a main title
fig.suptitle('Comparison of Population and Land Area Distribution', fontsize=20, fontweight='bold', y=1.05)

plt.tight_layout()
plt.show()
```

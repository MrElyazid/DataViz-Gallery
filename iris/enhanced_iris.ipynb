# Enhanced Iris Dataset Visualization

This notebook demonstrates enhanced visualization techniques for the Iris dataset using the custom utility functions.

```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Import our custom utility functions
from iris_utils import (
    load_iris_data,
    split_by_class,
    plot_feature_distributions,
    plot_feature_comparison,
    create_pairplot,
    plot_3d_scatter,
    IRIS_COLORS
)
```

## 1. Load and Explore the Dataset

```python
# Load the Iris dataset
df = load_iris_data('iris.data')

# Display the first few rows
print("First 5 rows of the Iris dataset:")
display(df.head())

# Display basic statistics
print("\nBasic statistics:")
display(df.describe())

# Display class distribution
print("\nClass distribution:")
display(df['class'].value_counts())
```

## 2. Split the Dataset by Class

```python
# Split the dataset by class
class_dfs = split_by_class(df)

# Access individual class DataFrames
setosa_df = class_dfs['Iris-setosa']
versicolor_df = class_dfs['Iris-versicolor']
virginica_df = class_dfs['Iris-virginica']

print(f"Number of Setosa samples: {len(setosa_df)}")
print(f"Number of Versicolor samples: {len(versicolor_df)}")
print(f"Number of Virginica samples: {len(virginica_df)}")
```

## 7. Correlation Analysis

```python
# Calculate correlation matrix
corr_matrix = df.drop('class', axis=1).corr()

# Create a heatmap of the correlation matrix
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='viridis', vmin=-1, vmax=1, 
            linewidths=0.5, cbar_kws={'label': 'Correlation Coefficient'})
plt.title('Correlation Matrix of Iris Features', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()

# Calculate correlation matrices for each class
print("\nCorrelation Matrix for Setosa:")
display(setosa_df.drop('class', axis=1).corr())

print("\nCorrelation Matrix for Versicolor:")
display(versicolor_df.drop('class', axis=1).corr())

print("\nCorrelation Matrix for Virginica:")
display(virginica_df.drop('class', axis=1).corr())
```

## 8. Summary Statistics by Class

```python
# Create a function to generate summary statistics
def class_summary(class_df, class_name):
    summary = class_df.describe().T
    summary['range'] = summary['max'] - summary['min']
    summary['coef_var'] = summary['std'] / summary['mean'] * 100  # Coefficient of variation in %
    return summary[['count', 'min', 'max', 'range', 'mean', 'std', 'coef_var']]

# Display summary statistics for each class
print("Summary Statistics for Setosa:")
display(class_summary(setosa_df.drop('class', axis=1), 'Setosa'))

print("\nSummary Statistics for Versicolor:")
display(class_summary(versicolor_df.drop('class', axis=1), 'Versicolor'))

print("\nSummary Statistics for Virginica:")
display(class_summary(virginica_df.drop('class', axis=1), 'Virginica'))
```

## 9. Feature Importance Analysis

```python
# Calculate the variance of each feature across classes
feature_means = {}
for feature in ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']:
    feature_means[feature] = {
        'Setosa': setosa_df[feature].mean(),
        'Versicolor': versicolor_df[feature].mean(),
        'Virginica': virginica_df[feature].mean()
    }

# Convert to DataFrame for easier visualization
means_df = pd.DataFrame(feature_means)

# Calculate the variance of means across classes (higher variance = better discriminator)
feature_variance = means_df.var()
feature_variance = feature_variance.sort_values(ascending=False)

# Plot the feature importance based on variance
plt.figure(figsize=(10, 6))
bars = plt.bar(feature_variance.index, feature_variance.values, color=sns.color_palette("viridis", 4))

# Add data labels
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + 0.01,
            f'{height:.2f}',
            ha='center', va='bottom', fontweight='bold')

plt.title('Feature Importance Based on Between-Class Variance', fontsize=16, fontweight='bold')
plt.xlabel('Feature', fontsize=14)
plt.ylabel('Variance of Class Means', fontsize=14)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

# Display the means by class
print("Feature Means by Class:")
display(means_df)

print("\nFeature Importance (Variance of Means):")
display(feature_variance)
```
